## 5. Cleaning Kolom Height & Weight

- **Kolom asal:** `Height`, `Weight`
- **Kolom baru:** `height_cm` (int), `weight_kg` (int)
- **Strategi:** Regex parse → konversi satuan → validasi range wajar

```python
def parse_height_ft_in(height_str):
    """
    Konversi string '5'7"' atau '5'11"' ke cm (integer).
    Return None jika gagal parse.
    """
    if pd.isna(height_str):
        return None
    # Regex: capture feet dan inches
    match = re.match(r"(\d+)'(\d+)\"?", str(height_str).strip())
    if not match:
        return None
    feet = int(match.group(1))
    inches = int(match.group(2))
    cm = round(feet * 30.48 + inches * 2.54)
    return cm

# Test fungsi dengan data dummy
test_heights = ["5'7\"", "6'0\"", "5'11\"", "4'11\"", "6'2\""]
for h in test_heights:
    print(f"{h:>6} → {parse_height_ft_in(h)} cm")
```

```python
df_fifa21['height_cm'] = df_fifa21['Height'].apply(parse_height_ft_in)

print("Sample hasil:")
print(df_fifa21[['Height', 'height_cm']].head(10))
print(f"\nMissing height_cm: {df_fifa21['height_cm'].isna().sum()}")
print(f"Range: {df_fifa21['height_cm'].min()} - {df_fifa21['height_cm'].max()} cm")
```

### 5.1 Validasi `height_cm`

| Cek | Hasil |
|-----|-------|
| Missing values | 0 |
| Min height | 155 cm |
| Max height | 206 cm |
| Tipe data | `int64` |

**Contoh pemain terpendek/tertinggi:**
- Terpendek: `N. Barrios dan H. Nakagawa dengan tinggi masing-masing 155cm`
- Tertinggi: `T. Holý dengan tinggi 206cm`

```python
def parse_weight_lbs(weight_str):
    """
    Konversi string '159lbs' atau '159 lbs' ke kg (integer).
    Return None jika gagal parse.
    """
    if pd.isna(weight_str):
        return None
    # Hapus 'lbs'/'LBS'/'lbs.' dan spasi, ambil angka
    cleaned = re.sub(r'[^\d.]', '', str(weight_str))
    if not cleaned:
        return None
    lbs = float(cleaned)
    kg = round(lbs * 0.453592)
    return kg

test_weights = ["159lbs", "159 lbs", "200lbs", "165.5 lbs", "180LBS"]
for w in test_weights:
    print(f"{w:>10} → {parse_weight_lbs(w)} kg")
```

```python
df_fifa21['weight_kg'] = df_fifa21['Weight'].apply(parse_weight_lbs)

print("Sample hasil:")
print(df_fifa21[['Weight', 'weight_kg']].head(10))
print(f"\nMissing weight_kg: {df_fifa21['weight_kg'].isna().sum()}")
print(f"Range: {df_fifa21['weight_kg'].min()} - {df_fifa21['weight_kg'].max()} kg")
```

### 5.2 Validasi `weight_kg`

| Cek | Hasil |
|-----|-------|
| Missing values | 0 |
| Min weight | 50 kg |
| Max weight | 110 kg |
| Tipe data | `int64` |

**Contoh pemain terpendek/tertinggi:**
- Terringan: `B. Al Mutairi dengan berat badan 50 kg`
- Terberat: `A. Akinfenwa dengan berat badan 110 kg`

```python
def iqr_bounds(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return lower, upper

for col in ['height_cm', 'weight_kg']:
    lower, upper = iqr_bounds(df_fifa21[col])
    outliers = df_fifa21[(df_fifa21[col] < lower) | (df_fifa21[col] > upper)]
    print(f"{col}: {len(outliers)} outlier (bounds: {lower:.0f} - {upper:.0f})")
    if len(outliers) > 0:
        print(outliers[['Name', col]].head())
        print()
```

## 7. Cleaning Value, Wage & Release Clause

- **Kolom asal:** `Value`, `Wage`, `Release Clause` (string: `"€67.5M"`, `"€560K"`, `"€138.4M"`, `"€0"`)
- **Kolom baru:** `value_eur`, `wage_eur`, `release_clause_eur` (integer Euro)
- **Strategi:** Fungsi `parse_eur()` handle `K` (ribuan), `M` (jutaan), `B` (miliaran — kalau ada), dan nilai tanpa suffix

```python
def parse_eur(value_str):
    """
    Konversi string '€67.5M', '€560K', '€0' ke integer Euro.
    Return 0 jika '€0' atau gagal parse.
    """
    if pd.isna(value_str):
        return 0

    s = str(value_str).strip().upper()

    # Hapus simbol € dan spasi
    s = s.replace('€', '').replace('EUR', '').replace(' ', '')

    # Handle "0" atau "-"
    if s in ['0', '-', '']:
        return 0

    # Deteksi suffix
    multiplier = 1
    if s.endswith('M'):     # Million (juta)
        multiplier = 1_000_000
        s = s[:-1]
    elif s.endswith('K'):     # Thousand (ribu)
        multiplier = 1_000
        s = s[:-1]

    try:
        return int(float(s) * multiplier)
    except ValueError:
        return 0

# Test dengan data dummy
test_values = [
    "€67.5M", "€560K", "€138.4M", "€0", "€2.5M", "€800K",
    "€1.2B", "€45.00M", "€12K", "-", "€5.5M"
]

print("=== TEST parse_eur() ===")
for v in test_values:
    print(f"{v:>10} → {parse_eur(v):>12,} EUR")
```

```python
df_fifa21['value_eur'] = df_fifa21['Value'].apply(parse_eur)
df_fifa21['wage_eur'] = df_fifa21['Wage'].apply(parse_eur)
df_fifa21['release_clause_eur'] = df_fifa21['Release Clause'].apply(parse_eur)

# Cek hasil sample
print("=== SAMPLE HASIL ===")
cols_show = ['Name', 'Value', 'value_eur', 'Wage', 'wage_eur', 'Release Clause', 'release_clause_eur']
print(df_fifa21[cols_show].head(15).to_string(index=False))

# Statistik missing & zero
print(f"\n=== STATISTIK ===")
for col in ['value_eur', 'wage_eur', 'release_clause_eur']:
    zero_count = (df_fifa21[col] == 0).sum()
    print(f"{col}: zero = {zero_count:,} ({(zero_count/len(df_fifa21)*100):.1f}%), max = {df_fifa21[col].max():,} EUR")
```

## 8. Cleaning Hits Column

- **Kolom asal:** `Hits` (string: `"\n372\n"`, `"\r\n1250\r\n"`, `"45\n"`, dll.)
- **Kolom baru:** `hits` (integer)
- **Strategi:** `str.strip()` hapus whitespace/newline → `pd.to_numeric(errors='coerce')` → `fillna(0).astype(int)`

```python
print("=== CONTOH RAW HITS (repr) ===")
for i, val in enumerate(df_fifa21['Hits'].head(10)):
    print(f"  Row {i}: {repr(val)}")

# 2. Cleaning pipeline
df_fifa21['hits'] = (
    df_fifa21['Hits']
    .astype(str)
    .str.strip()
    .str.replace(',', '', regex=False)
    .pipe(pd.to_numeric, errors='coerce')
    .fillna(0)
    .astype(int)
)

# 3. Cek hasil
print("\n=== SAMPLE HASIL ===")
print(df_fifa21[['Name', 'Hits', 'hits']].head(15).to_string(index=False))

print(f"\n=== STATISTIK ===")
print(f"Missing original: {df_fifa21['Hits'].isna().sum()}")
print(f"Zero hits: {(df_fifa21['hits'] == 0).sum():,}")
print(f"Max hits: {df_fifa21['hits'].max():,}")
print(f"Mean hits: {df_fifa21['hits'].mean():,.0f}")
```
