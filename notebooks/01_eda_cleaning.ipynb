## 3. Analisis Format Kolom Bermasalah

### 3.1 Height
**Contoh nilai:** `5'7"`
**Pola:** `feet'inches"`
**Target:** Integer cm

### 3.2 Weight
**Contoh nilai:** `159lbs`
**Pola:** angka + `lbs`
**Target:** Integer kg

### 3.3 Joined
**Contoh nilai:** `Jul 1, 2004`
**Pola:** `Mon D, YYYY`
**Target:** 3 kolom (`joined_year`, `joined_month`, `joined_day`)

### 3.4 Value / Wage / Release Clause
**Contoh:** `€67.5M`, `€560K`, `€138.4M`
**Pola:** `€` + angka + `K`/`M`
**Target:** Integer EUR

### 3.5 Hits
**Masalah:** Ada karakter `\n` di awal
**Target:** Integer bersih

### 3.6 Team & Contract
**Contoh (repr):**
'FC Barcelona\n2004 ~ 2021\n'
**Pola:** `Team\nStart ~ End\n`
**Target:** Pisah menjadi:
- `Team` → `"FC Barcelona"`
- `Contract` → `"2004 ~ 2021"` lalu dipisahkan kembali menjadi `contract_start`, `contract_end`

## 4. Rencana Cleaning

| Urutan | Kolom | Metode | Output |
|--------|-------|--------|--------|
| 1 | Height | Parse feet/inches → cm | `height_cm` (int) |
| 2 | Weight | Parse lbs → kg | `weight_kg` (int) |
| 3 | Joined | `pd.to_datetime` → extract | `joined_year`, `joined_month`, `joined_day` |
| 4 | Value | Parse € + K/M → int | `value_eur` (int) |
| 5 | Wage | Parse € + K/M → int | `wage_eur` (int) |
| 6 | Release Clause | Parse € + K/M → int | `release_clause_eur` (int) |
| 7 | Hits | `str.strip()` → int | `hits` (int) |
| 8 | Team & Contract | Split `\n` → Team, Contract | `team`, `contract_start`, `contract_end` |

> **Catatan:** Kolom asli dipertahankan, kolom baru ditambahkan dengan suffix `_clean` atau nama baru.
